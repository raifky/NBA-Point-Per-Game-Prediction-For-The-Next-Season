Import Liblary

```python
import joblib
import json
import pandas as pd
```

Melakukan inference dengan data baru

```python
model = joblib.load('ridge_pipeline.pkl')

# load metadata
with open('model_metadata.json', 'r') as f:
    metadata = json.load(f)

required_cols = metadata['numerical_features'] + metadata['categorical_features']

# data baru
X_new = pd.DataFrame([{
    'Age': 36,
    'MP': 32.2,
    'FGA': 18.0,
    '3PA': 11.2,
    'FTA': 4.3,
    'AST': 6.0,
    'TRB': 4.4,
    'TOV': 2.9,
    'Pos': 'PG'
}])

# validasi kolom
missing = set(required_cols) - set(X_new.columns)
if missing:
    raise ValueError(f"Kolom hilang: {missing}")

# predict
pred = model.predict(X_new)
print("Prediksi Point untuk next season:", pred[0])
```

Setelah melakukan inference model, bahwa model berhasil predict data baru yang di berikan

Model regresi yang dibuat (menggunakan Ridge Regression Pipeline) mampu melakukan prediksi PTS dengan data baru.

Beberapa poin penting dari model inference:

Model berhasil menangkap pola linear antara fitur statistik pemain dan PTS

Ridge Regression membantu mengurangi risiko overfitting akibat multikolinearitas antar fitur

Hasil evaluasi dan cross-validation menunjukkan bahwa model memiliki generalization yang baik

Model cocok digunakan sebagai baseline prediction model untuk prediksi PTS pemain NBA

Namun, model ini masih memiliki keterbatasan dalam menangkap hubungan non-linear dan faktor kontekstual pertandingan.
